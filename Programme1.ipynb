```python
# Conversion de la colonne 'date_added' en type DateTime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Vérifier la conversion
print(df['date_added'].dtypes)
```

```python
# Examiner d'abord les valeurs uniques dans la colonne
print(df['duration'].unique())

# Créer une fonction pour extraire le nombre 
def extract_numeric(value):
    if isinstance(value, str):
        # Extraire seulement le nombre au début de la chaîne
        import re
        match = re.search(r'^\d+', value)
        if match:
            return float(match.group())
    return value

# Appliquer la fonction à la colonne
df['duration_numeric'] = df['duration'].apply(extract_numeric)

# Vérifier le résultat
print(df[['duration', 'duration_numeric']].head())
```

```python
import re

# Fonction pour extraire le nombre en début de chaîne
def extract_number(value):
    if isinstance(value, str):
        match = re.search(r'^\d+', value)
        if match:
            return float(match.group())
    return value

# Appliquer cette fonction à la colonne 'duration'
df['duration'] = df['duration'].apply(extract_number)

# Vérifier le résultat
print(df['duration'].dtypes)
print(df['duration'].head())
```

```python
import seaborn as sns

# Visualiser la répartition des types d'œuvres
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='type', palette='Set2')
plt.title('Répartition des types d\'œuvres')
plt.xlabel('Type')
plt.ylabel('Nombre')
plt.show()

# Visualiser la répartition des notes
plt.figure(figsize=(10, 5))
top_countries = df['country'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 des pays les plus représentés')
plt.xlabel('Nombre d\'œuvres')
plt.ylabel('Pays')
plt.show()

# Visualiser la répartition des notes
plt.figure(figsize=(12, 5))
sns.histplot(data=df, x='release_year', bins=30, kde=False, color='skyblue')
plt.title('Répartition des œuvres par année')
plt.xlabel('Année de sortie')
plt.ylabel('Nombre')
plt.show()

# Visualiser la répartition des notes
plt.figure(figsize=(12, 5))
sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index, palette='pastel')
plt.title('Répartition des ratings')
plt.xlabel('Classification')
plt.ylabel('Nombre')
plt.xticks(rotation=45)
plt.show()


# Filtrer les films uniquement
films = df[df['type'] == 'Movie'].copy()

# Supprimer les lignes avec valeurs manquantes
films = films[films['duration'].notna()]

# S'assurer que la colonne est bien de type string
films['duration'] = films['duration'].astype(str)

# Extraire les minutes depuis la chaîne (ex: "90 min" -> 90)
films['duration_min'] = films['duration'].str.extract(r'(\d+)').astype(float)

# Visualisation avec Seaborn
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(films['duration_min'], bins=30, kde=True, color='salmon')
plt.title("Répartition de la durée des films")
plt.xlabel("Durée (en minutes)")
plt.ylabel("Nombre de films")
plt.show()

```

```python
#rérépartition de la durée des films
# Filtrer uniquement les séries
series = films[films['type'] == 'TV Show'].copy()

# Nettoyer la colonne duration (ex: "1 Season" -> 1, "3 Seasons" -> 3)
series['duration_seasons'] = series['duration'].str.extract(r'(\d+)').astype(float)

plt.figure(figsize=(10, 5))
sns.histplot(series['duration_seasons'], bins=20, kde=True, color='pastel')
plt.title("Répartition de la durée des séries (en saisons)")
plt.xlabel("Nombre de saisons")
plt.ylabel("Nombre de séries")
plt.show()
```

```python
# Découper les genres et les exploser en lignes
genres = films['listed_in'].str.split(', ', expand=True).stack()
genre_counts = genres.value_counts().head(15)

plt.figure(figsize=(10, 5))
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='blues')
plt.title("Top 15 des genres d'œuvres")
plt.xlabel("Nombre d'œuvres")
plt.ylabel("Genre")
plt.show()
```
